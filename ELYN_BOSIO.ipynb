```python
import pandas as pd
import matplotlib.pyplot as p1t

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
print(iris.feature_names)
```

```python
print(iris.target_names)
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
df["species"] = iris.target_names[iris.target]
df.head()
```

```python
print("Number of rows: ", df.shape[0])
print("Number of Columns: ", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df["species"].unique())
```

```python
print(df["species"].value_counts())
```

```python
p1t.figure(figsize=(8,6))

p1t.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"]

)
p1t.xlabel("sepal length (cm)")
p1t.title("sepal length vs sepal width")

p1t.show()
```

```python
p1t.figure(figsize=(8,6))

for species in df["species"].unique():
  subset = df[df["species"]==species]

  p1t.scatter(
      subset["sepal length (cm)"],
      subset["sepal width (cm)"],
      label = species

  )

  p1t.xlabel("sepal length (cm)")
  p1t.ylabel("sepal width (cm)"),
  p1t.title("iris Species Based on Sepal Measurement")
  p1t.legend()

  p1t.show()
```

```python
x = iris.data
y= iris.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y

)

print("Training observations: ", len(x_train))
print("Testing observations: ", len(x_test))
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print(y_pred)
```

```python
print("Actual: ")
print(y_test)

print("\nPredicted: ")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: ", accuracy)
```

```python
new_flower = [[5.1, 3.5, 1.4, 0.2]]
```

```python
prediction=model.predict(new_flower)
```

```python
print(
    "Prediction: ",
    iris.target_names[prediction[0]]

)
```

```python
probabilities = model.predict_proba(new_flower)
print(probabilities)
```

```python
for species, probability in zip(
    iris.target_names,
    probabilities[0]
):
    print(f"{species}: {probability:2%}")
```

```python
k_values = [1,2,3,5,7,9]
accuracies =[]

for k in k_values:
  model = KNeighborsClassifier(n_neighbors=k)
  model.fit(x_train, y_train)
  y_pred = model.predict(x_test)
  accuracy = accuracy_score(y_test, y_pred)
  accuracies.append(accuracy)
  print(f"k={k}, Accuracy={accuracy:2f}")
```

```python
p1t.figure(figsize=(8,5))
p1t.plot(
    k_values,
    accuracies,
    marker="o"

)
p1t.xlabel("k Value")
p1t.ylabel("Accuracy")
p1t.title("KNN Model Accuracy for Different K values")

p1t.xticks(k_values)
p1t.show()
```

```python
results = pd.DataFrame({
    "Actual":iris.target_names[y_test],
    "Predicted":iris.target_names[y_pred]
})

results.head(10)
```
